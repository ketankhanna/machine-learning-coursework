**Notes on the original results**
1. About 78% of clients do *not* default, so a model that always says "no default" already scores ~78%. The 81-82% accuracies above are only a few points better than that baseline.
2. The random forest scores 99.9% on training data but 81.5% on test data: a clear sign of **overfitting** (fully grown trees memorise the training set).
3. The KNN cell prints "random forest" in its labels, and KNN actually matches the forest on test accuracy (81.6% vs 81.5%), so it is not weaker.
4. The SGD cross-validation ran on unscaled features, which hurts SGD badly.

A fairer comparison with a baseline, a regularised forest and ROC AUC:

```python
import os, numpy as np, pandas as pd
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.dummy import DummyClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import SGDClassifier
from sklearn.metrics import accuracy_score, roc_auc_score

df = pd.read_excel('default of credit card clients.xls', skiprows=1, header=0)
X = df.drop(columns=['ID', 'default payment next month']); y = df['default payment next month']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=123, stratify=y)
models = {
  'Baseline (always no)': DummyClassifier(strategy='most_frequent'),
  'Random forest (regularised)': RandomForestClassifier(n_estimators=300, min_samples_leaf=20, random_state=123, n_jobs=-1),
  'KNN (k=50, scaled)': make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=50)),
  'SGD (scaled)': make_pipeline(StandardScaler(), SGDClassifier(loss='log_loss', random_state=123)),
}
for name, m in models.items():
    m.fit(X_train, y_train)
    auc = roc_auc_score(y_test, m.predict_proba(X_test)[:, 1])
    cv = cross_val_score(m, X_train, y_train, cv=4, scoring='accuracy').mean()
    print(f'{name:<28} test acc {accuracy_score(y_test, m.predict(X_test)):.3f} | 4-fold CV acc {cv:.3f} | test AUC {auc:.3f}')
```

The regularised forest keeps the same accuracy without memorising the training data and has the best AUC, so it remains the best choice, but by ranking quality (AUC) rather than raw accuracy. Scaling lifts SGD to about 79-81% accuracy, although it still ranks clients worse (lower AUC) than the forest.

**Q2 (optional): data augmentation on MNIST.** Shift each training image one pixel in four directions, then retrain. (Uses a 20,000-image training sample and KNN so it runs in a few minutes.)

```python
from scipy.ndimage import shift
from sklearn.datasets import fetch_openml
Xm, ym = fetch_openml('mnist_784', version=1, return_X_y=True, as_frame=False, parser='auto')
Xtr, ytr, Xte, yte = Xm[:20000], ym[:20000], Xm[60000:], ym[60000:]

def shift_image(img, dx, dy):
    return shift(img.reshape(28, 28), [dy, dx], cval=0, mode='constant').reshape(-1)

aug_X, aug_y = [Xtr], [ytr]
for dx, dy in [(1, 0), (-1, 0), (0, 1), (0, -1)]:
    aug_X.append(np.apply_along_axis(shift_image, 1, Xtr, dx, dy)); aug_y.append(ytr)
Xaug, yaug = np.vstack(aug_X), np.concatenate(aug_y)

knn = KNeighborsClassifier(n_neighbors=4, weights='distance', n_jobs=-1)
base = accuracy_score(yte, knn.fit(Xtr, ytr).predict(Xte))
aug = accuracy_score(yte, knn.fit(Xaug, yaug).predict(Xte))
print(f'Original 20k images:        {base:.4f}')
print(f'Augmented to {len(Xaug):,} images: {aug:.4f}')
```

Adding shifted copies raises test accuracy by roughly a point with no new data collected, which is the point of the exercise: the model learns that a digit moved one pixel is still the same digit.
